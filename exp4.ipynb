## ***Part A: Normal Distribution***
***Generate and visualize Normal Distribution using NumPy and Matplotlib.***


```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Generate normal distribution data
mu = 0
sigma = 1
data = np.random.normal(mu, sigma, 1000)

# Plot histogram
plt.hist(data, bins=30, density=True, alpha=0.6, color='g')

# Plot PDF curve
x = np.linspace(min(data), max(data), 100)
plt.plot(x, stats.norm.pdf(x, mu, sigma), 'k')

plt.title("Normal Distribution")
plt.show()
```

## **Part B: Descriptive Statistics**
***Calculate Mean, Median, Variance, and Standard Deviation.***

```python
mean = np.mean(data)
median = np.median(data)
variance = np.var(data)
std_dev = np.std(data)

print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Standard Deviation:", std_dev)
```

## **Part C: Coin Toss Simulation**
***Simulate Coin Toss and Calculate Experimental Probabilities.***


```python
# Simulate 1000 coin tosses
tosses = np.random.choice(['H', 'T'], size=1000)

heads = np.sum(tosses == 'H')
tails = np.sum(tosses == 'T')

print("Heads:", heads)
print("Tails:", tails)
print("Probability of Heads:", heads/1000)
print("Probability of Tails:", tails/1000)

# Bar graph
plt.bar(['Heads', 'Tails'], [heads, tails])
plt.title("Coin Toss Simulation")
plt.show()
```
